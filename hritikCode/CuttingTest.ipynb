```python
# imports
import os
import sys
import requests
import json
import ffmpeg
from tqdm import tqdm
from scenedetect import detect, ContentDetector, split_video_ffmpeg
```

```python
# Function to split a video into clips using scenedetect
def split_video_into_clips(video_path):
    # Get video name
    video_name = os.path.splitext(os.path.basename(video_path))[0]
    output_dir = os.path.join(os.getcwd(), video_name)

    # Create a new folder
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)
    
    # Get scene timestamps
    scene_list = detect(video_path, ContentDetector())
    
    split_video_ffmpeg(video_path, scene_list, output_dir=output_dir)
```

```python
# Example usage:
video_path = "Videos/Boyhood.mp4"
split_video_into_clips(video_path)
```

```python
# List of video paths
local_video_paths = ["Videos/Past Lives.mp4", "Videos/Parasite.mp4", "Videos/Joker.mp4", "Videos/Boyhood.mp4", "Videos/Back To The Future.mp4"]


main_server_path = "/mnt/labdata/naturalistic_face_project/movies/Movies after conversion"
# get names of all movies in path
movies = os.listdir(main_server_path)
# get full path of all movies
server_video_paths = [os.path.join(main_server_path, movie) for movie in movies]


# Loop through each video and split into clips
for video_path in video_paths:
    split_video_into_clips(video_path)
```
